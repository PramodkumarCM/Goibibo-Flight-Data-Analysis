#
Converting Flight Date Column to Datetime Format 
To ensure accurate time-series analysis and enable robust date-based operations, the flight date column in the dataset was converted from string to a proper datetime object using the following command:

```python
df['flight date'] = pd.to_datetime(df['flight date'], format='%d-%m-%Y', errors='coerce')
```

#
In this part of the project, the departure (dep_time) and arrival times (arr_time) were originally represented as float or integer values corresponding to fractions of days. To make this data human-readable and more intuitive for analysis
The final dep_time and arr_time columns now display cleanly formatted 24-hour time values (e.g., 14:35), significantly improving readability and facilitating downstream tasks like visualizations, time-based filtering, and scheduling analysis.

```python
# Convert the float values to timedelta representing fraction of a day
df['dep_time'] = pd.to_timedelta(df['dep_time'], unit='D')

# Function to format timedelta as %H:%M
def format_timedelta_hm(td):
    hours = td.components.days * 24 + td.components.hours
    minutes = td.components.minutes
    return f"{hours:02}:{minutes:02}"

# Apply the formatting function to the timedelta column
df['dep_time'] = df['dep_time'].apply(format_timedelta_hm)

# Convert the float values to timedelta representing fraction of a day
df['arr_time'] = pd.to_timedelta(df['arr_time'], unit='D')

# Function to format timedelta as %H:%M
def format_timedelta_hm(td):
    hours = td.components.days * 24 + td.components.hours
    minutes = td.components.minutes
    return f"{hours:02}:{minutes:02}"

# Apply the formatting function to the timedelta column
df['arr_time'] = df['arr_time'].apply(format_timedelta_hm)
```

#
To enable accurate time-based analysis, the dep_time and arr_time columns—initially stored as string representations of hours and minutes (e.g., "14:30")—were converted to timedelta objects using Pandas' to_timedelta() function. By appending ':00', we standardized the format to include seconds (HH:MM:SS), ensuring compatibility with timedelta conversion. This transformation facilitates easy computation of flight durations, delays, and scheduling comparisons within the dataset.

# 
•	Which airline has the fastest flights between major cities?

```python
# Group by airline and route (from -> to), then calculate average duration
fastest_routes = (
    df.groupby(['airline', 'from', 'to'])['duration']
    .mean()
    .reset_index()
)

# Now, find the fastest airline for each city pair
fastest_airlines = (
    fastest_routes.sort_values('duration')
    .groupby(['from', 'to'])
    .first()
    .reset_index()
)

fastest_airlines.max()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

# Create route column
fastest_airlines['route'] = fastest_airlines['from'] + ' → ' + fastest_airlines['to']

plt.figure(figsize=(10, 6))
sns.barplot(
    data=fastest_airlines,
    y='route',
    x='duration',
    hue='airline',
    palette='deep'
)
plt.title('Fastest Airline for Each Route')
plt.xlabel('Average Duration (minutes)')
plt.ylabel('Route')
plt.tight_layout()
plt.show()
```

#
Fastest Airlines Across Major City Routes
To determine which airline offers the quickest service between major cities, we performed a grouped analysis on the average flight durations across all route
This analysis highlights the most efficient airline for each route, which can be valuable for both travelers optimizing travel time and stakeholders evaluating airline performance. The use of average duration ensures the results account for all flights on that route rather than isolated cases.

# 
•	Do red-eye flights (late-night departures) have lower prices?

```python
# First, convert 'dep_time' to datetime (assuming HH:MM format)
df['dep_time'] = pd.to_datetime(df['dep_time'], format='%H:%M', errors='coerce')

# Extract the hour
df['dep_hour'] = df['dep_time'].dt.hour

# Create red-eye flag (between 10 PM and 5 AM)
df['is_redeye'] = df['dep_hour'].apply(lambda x: 1 if pd.notnull(x) and (x >= 22 or x < 5) else 0)

# Compare average prices
avg_prices = df.groupby('is_redeye')['price'].min().reset_index()
avg_prices.columns = ['Is Redeye', 'Average Price']

print(f'Yes there is flight at lower price in red-eye flights:{avg_prices}')
```

```python
# Assuming df already has 'is_redeye' and 'price'
plt.figure(figsize=(10, 6))
sns.boxplot(data=df, x='is_redeye', y='price', palette='Set2')

# Enhancing the chart
plt.title('Flight Price Distribution: Red-Eye vs Non-Red-Eye Flights', fontsize=14)
plt.xlabel('Is Red-Eye Flight (1 = Yes, 0 = No)', fontsize=12)
plt.ylabel('Flight Price', fontsize=12)
plt.grid(True)
plt.tight_layout()
plt.show()
```

#
The analysis suggests that red-eye flights tend to have lower minimum prices compared to regular flights. This trend is evident both statistically and visually from the box plot, where red-eye flights generally show a lower price distribution

# 
•	Which cities have the most direct flight connections?

```python
# Filter direct flights (non-stop flights)
direct_flights_df = df[df['stops'].str.lower() == 'non-stop']

# Count of direct flights from each city
from_city_direct_counts = direct_flights_df['from'].value_counts()

# Count of direct flights to each city
to_city_direct_counts = direct_flights_df['to'].value_counts()

# Combine both if needed
total_direct_connections = from_city_direct_counts.add(to_city_direct_counts, fill_value=0).sort_values(ascending=False)
total_direct_connections
```

#
Cities with Most Direct Flight Connections
To identify the cities with the strongest air travel connectivity, we focused on non-stop flights by filtering the dataset
This analysis highlights the cities that serve as major aviation hubs with the highest number of direct flight connections. These hubs are crucial for seamless travel and indicate strategic importance in the air transportation network.

```python
df_sns = total_direct_connections.reset_index()
df_sns.columns = ['City', 'Total Direct Connections']

# Set theme
sns.set(style='whitegrid')

# Plot
plt.figure(figsize=(10, 6))
ax = sns.barplot(
    x='Total Direct Connections',
    y='City',
    data=df_sns,
    palette='crest'
)

# Add title and labels
ax.set_title('Total Non-Stop Flight Connections by City', fontsize=14)
ax.set_xlabel('Total Direct Connections')
ax.set_ylabel('City')

# Add data labels
for i in ax.patches:
    ax.text(i.get_width() + 100, i.get_y() + 0.4, int(i.get_width()), color='black')

plt.tight_layout()
plt.show()
```

#
•	Is there a correlation between flight duration and price?

```python
correlation = df['duration'].corr(df['price'])

print(f"The correlation between flight duration and price is: {correlation:.2f}")

# Interpretation of the correlation coefficient:
if correlation > 0.7:
    print("There is a strong positive correlation, suggesting that longer flights tend to have higher prices.")
elif 0.3 < correlation <= 0.7:
    print("There is a moderate positive correlation.")
elif 0.1 < correlation <= 0.3:
    print("There is a weak positive correlation.")
elif -0.1 <= correlation <= 0.1:
    print("There is a very weak or no linear correlation.")
elif -0.3 <= correlation < -0.1:
    print("There is a weak negative correlation.")
elif -0.7 <= correlation < -0.3:
    print("There is a moderate negative correlation.")
else:
    print("There is a strong negative correlation, suggesting that longer flights tend to have lower prices (this is less likely in typical scenarios).")
```

 Correlation Between Flight Duration and Price
